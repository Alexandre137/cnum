## Atividade 1
Faça na linguagem Python uma função que verifica se um número é perfeito.

Um número é perfeito se a soma dos divisores for igual a ele mesmo (ex: 6 = 1 + 2 + 3).

````python
def is_perfect(n: int) -> bool:
    if n < 1:
        return False

    sum_divisors = 0
    for i in range(1, n):
        if n % i == 0:
            sum_divisors += i

    return sum_divisors == n


def main():
    assert is_perfect(6) == True
    assert is_perfect(7) == False
    assert is_perfect(-1) == False


if __name__ == "__main__":
    main()
````

## Atividade 2
Faça na linguagem Python uma função que calcula o fatorial de um número.

O fatorial de um número é o produto do número por todos os antecessores positivos.

````python
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError("O número deve ser não negativo.")

    result = 1
    i = n
    while i > 1:
        result *= i
        i -= 1

    return result

def main():
    assert factorial(5) == 120
    assert factorial(0) == 1
    try:
        factorial(-1)
    except ValueError as error:
        assert str(error) == "O número deve ser não negativo."

if __name__ == "__main__":
    main()
````

## Atividade 3
Faça na linguagem Python uma função que verifica se um número é primo.

Um número que só pode ser dividido por um e por ele mesmo.

````python
def es_primo(numero):
    # Números menores ou iguais a 1 não são primos
    if numero <= 1:
        return False

    # Verifica divisores de 2 até a raiz quadrada do número
    for i in range(2, int(numero**0.5) + 1):
        if numero % i == 0:
            return False  # Encontrou um divisor, então não é primo

    return True  # Não encontrou divisores, é primo
````

## Atividade 4
Faça na linguagem Python uma função que calcula a soma dos dígitos.

Caso número negativo, dispare uma mensagem de erro (ex: 123, 1 + 2 + 3 = 6).

````python
def soma_digitos(numero):
    # Se o número for negativo, lança um erro (ValueError)
    if numero < 0:
        raise ValueError("O número não pode ser negativo!")

    # Converte o número para texto, depois cada caractere em inteiro e soma tudo
    return sum(int(digito) for digito in str(numero))


# --- Exemplos de uso ---
print(soma_digitos(123))  # Retorna: 6 (1 + 2 + 3)
print(soma_digitos(905))  # Retorna: 14 (9 + 0 + 5)
print(soma_digitos(0))    # Retorna: 0

# Testando o caso de erro:
try:
    print(soma_digitos(-123))
except ValueError as e:
    print(f"Erro capturado: {e}")
````

Escreva um algoritmo que realize o seguinte produto matricial:

[ [1, 2, 3],  
  [4, 5, 6],  
  [7, 8, 9] ]  

multiplicado por:

[ [11, 12, 13],  
  [14, 15, 16],  
  [17, 18, 19] ]

````python
def multiplicar_matrizes(matriz_a, matriz_b):
    linhas_a = len(matriz_a)
    colunas_a = len(matriz_a[0])
    linhas_b = len(matriz_b)
    colunas_b = len(matriz_b[0])

    # Validação de dimensão para multiplicação matricial
    if colunas_a != linhas_b:
        raise ValueError("O número de colunas da Matriz A deve ser igual ao número de linhas da Matriz B.")

    # Cria matriz resultado preenchida com zeros (dimensão: linhas_a x colunas_b)
    resultado = [[0 for _ in range(colunas_b)] for _ in range(linhas_a)]

    # Multiplicação: linha por coluna
    for i in range(linhas_a):
        for j in range(colunas_b):
            for k in range(colunas_a):
                resultado[i][j] += matriz_a[i][k] * matriz_b[k][j]

    return resultado


# Matrizes fornecidas
A = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

B = [
    [11, 12, 13],
    [14, 15, 16],
    [17, 18, 19]
]

# Execução e exibição do resultado
resultado = multiplicar_matrizes(A, B)

print("Resultado do produto matricial:")
for linha in resultado:
    print(linha)
````
